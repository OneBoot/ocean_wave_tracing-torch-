```python
rays = torch.tensor([[1,2,3,4], [torch.nan, 4, 5, 6]])

indexes = torch.tensor([1,2])

print(rays)

Nanfilter = [torch.isnan(ray).any() for ray in rays]

nanfilter2 = torch.isnan(rays).any(dim=1)

print(Nanfilter)
print(indexes[nanfilter2])
```
